## 📖 핵심 개념

### 1) AI, 머신러닝, 딥러닝 — 뭐가 다를까?

이 세 단어는 뉴스에서 거의 같은 의미로 섞여 쓰이지만, 사실은 **크기가 다른 원 세 개**로 이해하면 가장 명확합니다.

```
┌─────────────────────────────┐
│   AI (인공지능)                │
│  ┌─────────────────────┐     │
│  │  머신러닝(ML)          │     │
│  │  ┌───────────────┐   │     │
│  │  │  딥러닝(DL)     │   │     │
│  │  └───────────────┘   │     │
│  └─────────────────────┘     │
└─────────────────────────────┘
```

- **AI**: 사람처럼 사고·행동하는 시스템 전체를 가리키는 가장 넓은 개념. 규칙을 사람이 일일이 정해줬든, 스스로 학습했든 상관없이 전부 AI로 분류됨
- **머신러닝(ML)**: 규칙을 사람이 정해주는 대신, 데이터를 잔뜩 보여주고 컴퓨터가 스스로 규칙(패턴)을 찾아내게 하는 방법
- **딥러닝(DL)**: 머신러닝 중에서도 사람 뇌의 신경세포(뉴런) 구조를 흉내 낸 "인공신경망"을 사용하는 방법. ChatGPT 같은 생성형 AI, 얼굴 인식 등이 대부분 여기에 해당

| 구분 | 정의 | 비유 | 예시 |
|---|---|---|---|
| **AI** | 사람처럼 사고·행동하는 시스템 전체 | 목적지 | 규칙 기반 챗봇, 게임 AI, 추천 시스템 |
| **머신러닝** | 데이터로 스스로 규칙을 학습 | 가는 방법 중 하나 | 스팸 필터, 이상거래 탐지 |
| **딥러닝** | 신경망 기반 머신러닝 | 그 방법 중 특정 교통수단 | 이미지 인식, LLM |

> 💡 **한 줄 요약**: AI는 목표, 머신러닝은 방법, 딥러닝은 그 방법 중 하나입니다.

### 2) 머신러닝의 학습 방식 3가지

- **지도학습(Supervised Learning)**: 정답(라벨)이 있는 데이터로 학습
- **비지도학습(Unsupervised Learning)**: 정답 없이 데이터의 숨은 패턴이나 그룹을 스스로 찾는 방식
- **강화학습(Reinforcement Learning)**: 행동의 결과로 보상/벌점을 받으며 점점 더 나은 행동을 학습하는 방식

| 학습 방식 | 정답 데이터 | 대표 예시 |
|---|---|---|
| 지도학습 | 있음 | 스팸 메일 분류, 집값 예측 |
| 비지도학습 | 없음 | 고객 세분화, 이상치 탐지 |
| 강화학습 | 보상/벌점 | 알파고, 자율주행 |

### 2) 변수 이름 짓는 규칙

- 공백 포함 (`my score`)
- 예약어(`class`, `for`, `if`, `True`, `False`, `None`, `import` 등)



### 3) 자료형이란 무엇일까?

###

같은 연산자(`+` 등)라도 값의 종류에 따라 전혀 다르게 동작합니다. 예를 들어 `10 + 5`는 숫자를 더해 `15`가 되지만, `"10" + "5"`는 문자열을 이어 붙여 `"105"`가 됩니다. 이렇게 **값이 어떤 종류인지를 구분하는 속성**을 자료형(Data Type)이라고 합니다.

**파이썬은 동적 타이핑(Dynamic Typing) 언어**

- 다른 언어(Java, C 등)는 `int x = 10;`처럼 변수를 만들 때 자료형을 미리 못 박아야 함
- 파이썬은 `x = 10`이라고만 써도 자동으로 int로 인식
→ 이런 방식을 **동적 타이핑**이라고 함
- 장점: 코드가 짧고 빠르게 작성 가능
- 단점: 실행해보기 전까지는 자료형 오류를 알기 어려움 (런타임 오류)

```

```

**4-3) 불리언형 — bool**

- 참(`True`) 또는 거짓(`False`) 딱 두 가지 값만 존재
- 비교 연산자(`>`, `==` 등)의 결과로 자주 만들어짐


`bool()` 진위값 판단 규칙:

| 값 | 결과 | 이유 |
|---|---|---|
| `0`, `0.0` | False | 숫자 0은 항상 False |
| 그 외 숫자 | True | 0이 아니면 전부 True |
| `""` | False | 내용이 없음 |
| `"0"`, `"hello"` | True | ⚠️ 내용이 있으므로 True (숫자 0과 다름!) |
| `[]`, `{}`, `()` | False | 내용이 없음 |
| `None` | False | 값이 아예 없음 |

<details>

### 7) 서로 다른 형변환 방식 비교 — 암묵적(자동) vs 명시적(수동)

자료형이 다른 값끼리 연산할 때, 파이썬이 **자동으로** 형을 맞춰주는 경우와 **직접** 형을 맞춰야 하는 경우가 있습니다. 이 둘을 구분하는 것이 중요합니다.

#### 암묵적 형변환 (Implicit Conversion) — 파이썬이 알아서 처리

- `int`와 `float`을 함께 연산하면, 정밀도 손실을 막기 위해 파이썬이 자동으로 `int`를 `float`으로 바꿔서 계산
- 사람이 별도로 형변환 함수를 쓰지 않아도 됨

<details>
<summary> ---------- 💡 정답 코드 보기 ---------- </summary>

```
# 암묵적 변환
result = 10 + 3.5
print(result, type(result))   # 13.5 <class 'float'>

```

</details>

````python
# 코드 입력

result = 10 + 3.5
print(result)
type(result)
````

#### 명시적 형변환 (Explicit Conversion) — 사람이 직접 지시

- `str`과 `int`처럼 파이썬이 자동으로 규칙을 정할 수 없는 조합은 **직접** `int()`, `str()` 등을 호출해야 함
- 이걸 생략하면 에러가 발생

<details>
<summary> ---------- 💡 정답 코드 보기 ---------- </summary>

```

# print(age_str + 1)      # ❌ TypeError
print(int(age_str) + 1)    # ✅ 26

```

</details>

````python
# 코드 입력

age_str = '25'
# print(age_str + 1) # 에러!
print(int(age_str) + 1)
````

| 명시적 변환 | str↔숫자처럼 자동 규칙이 없을 때 | `int("25")+1` | `int()`/`str()` 직접 호출 |

# 📘 CH02-03. 연산자

### 🤔 먼저 생각해보기
**편의점에서 "만 19세 이상이면서 신분증을 가지고 있어야 술을 살 수 있다"는 규칙, 코드로 어떻게 표현할 수 있을까요?**

"나이가 19살 이상"과 "신분증이 있다"라는 두 가지 조건을 각각 확인한 뒤, 이 둘을 **동시에** 만족해야 합니다. 여러 값을 비교하고, 그 결과들을 조합해서 하나의 참/거짓 판단으로 만드는 도구가 **연산자**입니다.

## 📖 핵심 개념

### 1) 산술 연산자

- `+` 덧셈,  뺄셈,  곱셈은 일상적인 사칙연산과 동일
- `/`: 나눗셈, 결과는 **항상 float** (딱 나누어떨어져도 예외 없음)
- `//`: 몫 연산, 소수점 이하를 버린 정수 나눗셈
- `%`: 나머지 연산
- `**`: 거듭제곱

<details>
<summary> ---------- 💡 정답 코드 보기 ---------- </summary>

```
print(7 + 3)   # 10
print(7 - 3)   # 4
print(7 * 3)   # 21
print(7 / 3)   # 2.333... (나눗셈은 항상 float)
print(7 // 3)  # 2   몫
print(7 % 3)   # 1   나머지
print(7 ** 2)  # 49  거듭제곱

```

</details>

````python
# 코드 입력

print(7 + 3)
print(7 - 3)
print(7 * 3)
print(7 / 3)  # 나눗셈은 항상 float
print(7 // 3) # 몫
print(7 % 3)  # 나머지
print(7 ** 3)

print(10 / 2) # int끼리 나눠도 결과는 float
````

> 💡 `//`와 `%`는 실무에서 자주 쓰입니다. "물건 17개를 5개씩 상자에 담으면?"은 `17//5`(몫=3상자)와 `17%5`(나머지=2개)로 정확히 표현됩니다.

### 6) 식별 연산자 (is / is not)
- `is`: 두 변수가 **메모리상 완전히 같은 객체**를 가리키는지 확인 (값이 같은지가 아님!)
- `==`는 “값이 같은가”, `is`는 “완전히 동일한 존재인가”를 묻는다는 점이 핵심 차이

<details>
<summary> ---------- 💡 정답 코드 보기 ---------- </summary>

```
a = [1, 2, 3]
b = [1, 2, 3]
c = a

print(a == b)   # True   (내용물 값이 같음)
print(a is b)   # False  (서로 다른 별개의 리스트)
print(a is c)   # True   (c는 a와 완전히 같은 대상)

```

</details>

````python
# 코드 입력

a = [1,2,3]
b = [1,2,3]
c = a

print(a == b)
print(a is b) # a와 b는 값은 같지만 주소값은 다름! 다른 존재이다.
print(a is c) # c = a 라고 저장했기 때문에 둘은 완전히 같다.
````

### 7) 서로 다른 연산자 비교

**`==` vs `is`**

| 연산자 | 묻는 질문 | 결과가 다를 수 있는 경우 |
|---|---|---|
| `==` | 값이 같은가? | 다른 리스트라도 내용이 같으면 True |
| `is` | 완전히 같은 객체인가? | 내용이 같아도 별개 객체면 False |

**논리 연산자(and/or) vs 멤버십 연산자(in)**
- `and`/`or`는 여러 조건식(True/False)을 하나로 합칠 때
- `in`은 하나의 값이 목록/문자열 안에 있는지 확인할 때

<details>
<summary> ---------- 💡 정답 코드 보기 ---------- </summary>

```
age = 20
fruit = "포도"
basket = ["사과", "바나나", "포도"]

# and: 조건 두 개를 합침
print(age >= 19 and fruit in basket) # True and True → True

# in 하나만 쓸 때
print(fruit in basket) # True

```

</details>

````python
# 코드 입력

age = 20
fruit = '포도'
basket = ['사과', '바나나', '포도']

print(age >= 19 and fruit in basket) # True and True -> True
print(fruit in basket)
````

# 📘 CH02-04. 조건문



### 🤔 먼저 생각해보기
**ATM은 어떻게 잔액이 부족하면 출금을 막을 수 있을까요?**

우리는 일상생활에서 수많은 '판단'을 합니다. 잔액이 충분하면 출금하고, 부족하면 출금을 막습니다. 이처럼 컴퓨터도 조건에 따라 다른 동작을 수행할 수 있는데, 이를 **조건문(Condition)**이라고 합니다.

## 📖 핵심 개념

### 1) 조건문이란? — 왜 필요한가

조건문은 프로그램의 흐름을 제어하는 구조로, **특정 조건이 참(True)일 때에만** 특정 코드가 실행되도록 만듭니다. 이를 통해 프로그램이 상황에 맞게 다르게 동작할 수 있습니다.

**실생활 예시로 이해하기**

- **신호등 제어**: 빨간 불이면 “멈춤”, 그렇지 않으면 “출발”
- **스테이크 굽기 판정**: 피가 보이면 “레어”, 피가 안 보이지만 겉을 태우면 “웰던”, 피도 안 보이고 겉도 태우지 않으면 “미디움”

이처럼 우리는 일상에서도 끊임없이 조건에 따라 다른 행동을 선택합니다. 조건문은 이 “만약 ~라면 ~하고, 아니면 ~한다”는 판단을 코드로 그대로 옮기는 도구입니다.

### 2) 조건문의 종류 — `if`, `elif`, `else` 각각 뜯어보기

파이썬 조건문은 세 가지 키워드로 구성됩니다. 각각의 역할이 명확히 다르므로 하나씩 따로 이해하는 것이 중요합니다.

#### ✅ `if` — 하나의 조건을 검사

- 조건 하나를 검사해서, **참(True)이면 실행**, **거짓(False)이면 실행하지 않음**
- `else` 없이 `if`만 단독으로 사용 가능 (거짓일 때 아무것도 안 해도 될 경우)

<details>
<summary> ---------- 💡 정답 코드 보기 ---------- </summary>

```
light = "red"
if light == "red":
    print("멈춰요")
# else가 없으므로 red가 아니면 아무것도 출력되지 않음

```

</details>

````python
# 코드 입력

light = 'red'

if light == 'red':
    print('멈춰요')
````

**else — 위 조건이 거짓일 때**

- `if`(또는 마지막 `elif`)의 조건이 거짓일 때 실행되는 블록
- 생략 가능 (조건이 거짓일 때 딱히 할 일이 없다면)

<details>
<summary> ---------- 💡 정답 코드 보기 ---------- </summary>

```

    print("멈춰요")
else:
    print("출발해요")

```

</details>

````python
# 코드 입력

light = 'green'
if light == 'red':
    print('멈춰요')
else:
    print('출발해요')
````

**elif — 다중 조건을 순차적으로 검사**

- `else if`의 축약형
- 여러 조건을 순서대로 검사할 때 사용하며, **앞선 조건이 모두 거짓일 때만** 다음 `elif` 조건을 평가
- 조건이 n개면 선택지는 최대 n+1개(마지막 `else` 포함)가 됨
- 참인 조건을 하나 만나 실행되면, **그 이후의 elif/else는 전부 무시**됨

<details>
<summary> ---------- 💡 정답 코드 보기 ---------- </summary>

```
blood = True

else:
    print("미디움")

```

</details>

💡 핵심 요약: if는 조건 검사의 시작, elif는 앞 조건이 거짓일 때 이어지는 추가 검사, else는 모든 조건이 거짓일 때의 마지막 처리입니다.

#### 5-2) 중첩 `if` vs `elif`

- **중첩 `if`**: “A이면서 동시에 B”처럼 **여러 조건을 AND로 동시에 만족**하는지 판단할 때 사용 (바깥 조건이 참일 때만 안쪽 조건을 검사)
- **`elif`**: 서로 배타적인 여러 조건 중 **딱 하나를 선택**할 때 사용

<details>
<summary> ---------- 💡 정답 코드 보기 ---------- </summary>

```
a = 15
if a % 2 == 0:

else:
    print("홀수")
# 결과: 홀수

```

</details>

````python
# 코드 입력

a = 20
if a % 2 == 0:
    if a < 10:
        print('짝수이면서 10보다 작은 수')
    else:
        print('짝수이면서 10보다 큰 수')
else:
    print('홀수')
````

| 구조 | 용도 | 실행 방식 |
|---|---|---|
| 중첩 if | 복합 조건(AND) — "OOO이면서 OOO" | 바깥 조건 참일 때 안쪽 검사 |
| elif | 배타적 조건 중 하나 선택 | 가장 먼저 참인 것 하나만 실행 |

- 조건에 따라 변수에 값을 바로 대입하고 싶을 때 유용
- 다만 조건이 복잡해지면 가독성이 떨어지므로, 그럴 땐 기본 `if/elif/else` 형태를 사용하는 것을 권장

## 📌 조건문 구조 요약표

| 구분 | 설명 | 실행 방식 |
|---|---|---|

| 독립 if 여러 개 | 서로 무관한 조건 각각 처리 | 만족하는 조건 전부 실행 |

### 6) break vs continue

- **`break`**: 조건을 만나면 반복문을 **완전히 종료**하고 빠져나감
- **`continue`**: 조건을 만나면 **이번 차례만 건너뛰고** 다음 반복으로 넘어감

| 구분 | 동작 | 이후 진행 |
|---|---|---|
| break | 반복문 자체를 완전히 종료 | 이후 반복 전혀 실행 안 됨 |
| continue | 이번 차례만 건너뜀 | 다음 차례부터 계속 진행 |
