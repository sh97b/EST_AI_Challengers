### 문제 48. (심화, 응용) while + 조건문 + break — 로그인 시도 제한
`correct_pw="1234"`, `attempts=["0000", "1111", "1234"]`가 주어집니다.
최대 3번까지 시도하고, 맞으면 `"로그인 성공"` + break, 3번 모두 틀리면 `"계정 잠금"`을 출력하세요.

```python
correct_pw = "1234"
attempts = ["0000", "1111", "1234"]
tries = 0
max_tries = 3
login_success = False

# while tries < max_tries: 조건으로 반복
#   pw_input = attempts[tries] 로 현재 시도값 가져오기
#   tries += 1
#   pw_input == correct_pw 이면: 성공 메시지 출력 → login_success = True → break
#   아니면: 실패 메시지 + 남은 기회 출력
# 반복 종료 후: login_success가 False면 "계정 잠금" 출력
while tries < max_tries:
  pw_input = attempts[tries]
  tries += 1
  if pw_input == correct_pw:
    print("로그인 성공")
    login_success = True
    break
  else:
    print(f'{tries}번째 시도 실패. 남은 기회: {max_tries - tries}')
if not login_success:
  print("계정 잠금")
```

### 문제 49. (심화, 종합) 편의점 포인트 적립 시스템
`payments` 리스트의 각 결제 금액에 대해 아래 규칙으로 포인트를 적립하고 출력하세요.

**규칙**
- 5,000원 미만 → 적립 없음 (earned = 0)
- 5,000원 이상 ~ 10,000원 미만 → 금액의 3% 적립 (`int(amount * 0.03)`)
- 10,000원 이상 → 금액의 5% 적립 (`int(amount * 0.05)`)
- 누적 포인트가 1,000 초과 시 `"🎉 포인트 사용 가능!"` 추가 출력

```python
payments = [3000, 8000, 15000, 4500, 12000, 9000]
total_points = 0

# for i, amount in enumerate(payments, start=1): 로 순회 (i는 1부터 시작)
#
# 1단계 — 적립 포인트(earned) 계산:
#   amount < 5000         → earned = 0
#   amount < 10000        → earned = int(amount * 0.03)
#   그 외                 → earned = int(amount * 0.05)
#
# 2단계 — 누적: total_points += earned
#
# 3단계 — 출력:
#   f"[{i}번째 결제] 금액: {amount:,}원 → 적립: {earned}점 (누적: {total_points}점)"
#   total_points >= 1000 이면 같은 줄에 "  🎉 포인트 사용 가능!" 추가 (end="" 활용)
#
# 반복 종료 후: f"\n최종 누적 포인트: {total_points}점" 출력

for i, amount in enumerate(payments, start=1):
  if amount < 5000:
    earned = 0
  elif amount < 10000:
    earned = int(amount * 0.03)
  else:
    earned = int(amount * 0.05)
  total_points += earned
  print(f'[{i}번째 결제] 금액: {amount:,}원 → 적립: {earned}점 (누적: {total_points}점)', end='')
  if total_points >= 1000:
    print("  🎉 포인트 사용 가능!")
  else:
    print()
print(f'\n최종 누적 포인트: {total_points}점')
```

`team_data` 리스트의 각 팀원에 대해 벌점을 계산하고, 평균보다 높은 팀원을 호출하세요.

- 지각 3회 이상이면 `"⚠️ 주의 대상"` 추가 표시
- 전체 평균 벌점 계산 후, 평균 초과자만 `"📢 이름 호출 필요"` 출력

```python
team_data = [
    {"name": "김민준", "late_count": 1},
    {"name": "이서연", "late_count": 4},
    {"name": "박지훈", "late_count": 0},
    {"name": "최유나", "late_count": 3},
    {"name": "정하늘", "late_count": 2},
]
penalty_per_late = 2

# 1단계: for member in team_data: 순회하며 벌점 계산
#   penalty = member["late_count"] * penalty_per_late
#   member["penalty"] = penalty  (딕셔너리에 벌점 저장)
#   total_penalty += penalty  (누적합)
#   late_count >= 3 이면 warning = " (⚠️ 주의 대상)", 아니면 ""
#   f"{member['name']}: 지각 {member['late_count']}회 → 벌점 {penalty}점{warning}" 출력
total_penalty = 0

for member in team_data:
  penalty = member["late_count"] * penalty_per_late
  member["penalty"] = penalty
  total_penalty += penalty
  if member["late_count"] >= 3:
    warning = " (⚠️ 주의 대상)"
  else:
    warning = ""
  print(f"{member['name']}: 지각 {member['late_count']}회 → 벌점 {penalty}점{warning}")

# 2단계: 평균 벌점 계산
#   average_penalty = total_penalty / len(team_data)
#   f"\n팀 평균 벌점: {average_penalty:.1f}점" 출력

average_penalty = total_penalty / len(team_data)
print(f"\n팀 평균 벌점: {average_penalty:.1f}점")

# 3단계: 평균 초과자만 출력
#   for member in team_data: 다시 순회
#   member["penalty"] > average_penalty 이면
#   f"📢 {member['name']} 호출 필요 (벌점 {member['penalty']}점)" 출력

for member in team_data:
  if member["penalty"] > average_penalty:
    print(f"📢 {member['name']} 호출 필요 (벌점 {member['penalty']}점)")
```
